# Company number

**Director names** (Companies House) and **managing director** (parsed from the **company website** only, plus optional `site:` search) are built in **`director_enrichment.ipynb`**. Run that notebook end-to-end; it reads **`Market Research List.csv`** and writes **`market_research_directors.csv`**.

To **look up company numbers from business names** using the Companies House search API, use **`fetch_company_numbers.ipynb`**.
